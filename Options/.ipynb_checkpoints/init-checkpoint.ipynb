```python
def get_sp_tickers():
    sp_tickers=pd.read_csv('sp500_tickers.csv')
    sp_tickers_flat=[i[0] for i in sp_tickers.values.tolist()]
    return sp_tickers_flat
```

```python
def get_options_dates(t):

    options={}
    if isinstance(t, list):
        for i in t:
            T=yf.Ticker(i)
            options[i]=T.options
    elif isinstance(t, str):
        T=yf.Ticker(i)
        options[i]=T.options
        
    return options
```

```python
def cherry_pick(t, n=10):
    picked=[]
    
    rng = np.random.default_rng()
    cherry_picker = rng.integers(0, len(t), size=n)
    
    for i in cherry_picker:
        picked.append(t[i])
    
    return picked
```

```python
def black_scholes_call(S, K, T, sigma, r=0.1):
    d1 = (np.log(S/K) + (r + (sigma**2)/2)*T) / (sigma*np.sqrt(T))
    d2 = d1 - sigma*np.sqrt(T)

    return (
        S*sci.norm.cdf(d1)
        - K*np.exp(-r*T)*sci.norm.cdf(d2)
    )

def black_scholes_put(S, K, T, sigma, r=0.03):
    d1 = (np.log(S/K) + (r + (sigma**2)/2)*T) / (sigma*np.sqrt(T))
    d2 = d1 - sigma*np.sqrt(T)

    return (
        K*np.exp(-r*T)*sci.norm.cdf(-d2)
        -S*sci.norm.cdf(-d1)
    )

def C_Greeks(S, K, T, sigma, r=0.03):
    d1 = (np.log(S/K) + (r + sigma**2/2)*T) / (sigma*np.sqrt(T))
    d2 = d1 - sigma*np.sqrt(T)
    DRV=sci.norm.pdf(d1)

    Delta=sci.norm.cdf(d1)
    
    Theta=(-r*K*np.exp(-r*T)*sci.norm.cdf(d2)
            -(sigma*S/(2*T))*DRV)
    
    Gamma=DRV/(sigma*S*np.sqrt(T))
    
    Vega=S*np.sqrt(T)*DRV
    
    Rho=K*T*np.exp(-r*T)*sci.norm.cdf(d2)

    return (
        (Delta,Theta,Gamma,Vega,Rho)
    )
```

```python
def get_returns(close_prices):
    returns=close_prices.pct_change()
    return returns
```

```python
def get_rolling_summary(returns, roll=30):
    rolling_dev=returns.rolling(roll).std()*np.sqrt(days_per_year)
    rolling_mean=returns.rolling(roll).mean()
    return (rolling_dev, rolling_mean)
```

```python
def delta_t_exp(dates):
    t_deltas=[]
    for j in dates:
        delta_t=round((dt.datetime.strptime(j, date_format)-now).days/days_per_year, 4)
        t_deltas.append((j, delta_t))

    return t_deltas
```

```python
def chain_collect(t, date):
    chain=yf.Ticker(t).option_chain(date)
            
    calls=chain.calls
    putz=chain.puts
    
    return (date, calls, putz)
```

```python
def date_strike(date_dict):
    date_strike_call=pd.DataFrame()
    date_strike_put=pd.DataFrame()
    
    for i in date_dict.keys():
        for e in date_dict[i]:
            d, call, put=chain_collect(i, e)
            c=call['strike']
            p=put['strike']
    
            date_strike_call[e]=c
            date_strike_put[e]=p
    return (date_strike_call, date_strike_call)
```
